# 8. Feature Engineering

####   1. Creating a New Columns that answers the Question "How many Days did the patient Stay"

```python
df["Lenght_of_Stay"] = (df["Discharge_Date"] - df["Date_of_Admission"]).dt.days 
```

```python
df =df.rename(columns={
    "Lenght_of_Stay" : "Days_Stayed"
})
```

```python
df.describe()
```

#### 2. Creating a new Flag Column based on the Column "Billing_Amount"

```python

conditions = [

    df["Billing_Amount"] > 30000,
    (df["Billing_Amount"] > 10000) & (df["Billing_Amount"] < 30000),
    df["Billing_Amount"] < 10000
]

choices = ["High Fees", "Medium Fees", "Low Fees"]

df["Billing_Category"] = np.select(conditions, choices, default="Unknown")
```

```python
df["Billing_Category"]
```

```python
unknown = df[df["Billing_Category"] == "Unknown"]

unknown
```

#### 3. Creating a New "Age_Category" column based on Column "Age"

```python
age_conditions = [

    df["Age"] < 18,
    (df["Age"] >= 18) & (df["Age"] <= 35),
    (df["Age"] >= 36) & (df["Age"] <= 50),
    (df['Age'] >= 51) & (df['Age'] <= 65),
    df['Age'] > 65   
]

age_choices = ['Pediatric',
    'Young Adult',
    'Middle-Aged Adult',
    'Pre-Senior',
    'Geriatric']


df["Age_Category"] = np.select(age_conditions, age_choices, default="n/a")
```
